# Set the data location

The folder every run of this microscope is saved under, for example `D:\ZMART-data`: a full path to an existing folder on a **local disk** this computer can write to. Copy the data to a network share after the run, not during it.

Set once, like the limits and the orientation; the driver loads it at every connect. A session or a single capture can still choose another folder (`output_root`, `save_to`).

```python
import bootstrap  # noqa: F401 -- puts the repository on the import path

from zmart_drivers.zeiss.light_microscope.zen_api_grpc.driver_setup.data_location import (
    data_location,
    set_data_location,
)

print("now:", data_location())
```

## Choose the folder

Write it below and run the cell. A folder that cannot hold the data is refused, with the reason.

```python
set_data_location(r"D:\ZMART-data")
print("set:", data_location())
```
