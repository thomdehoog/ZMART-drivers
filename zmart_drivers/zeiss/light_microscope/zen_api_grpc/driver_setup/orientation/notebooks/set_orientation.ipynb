# Measure the image orientation and the pixel size

The driver takes a snap, moves the stage 15 µm along x and takes another, then 15 µm along y and a third, and returns. How the picture shifted tells how the camera is turned against the stage and how large one pixel is.

Before you run it: publish the limits first, load an experiment whose snap shows the sample well (one channel is enough), and put a sample with structure (cells, beads, a grid) in focus under the objective. The snaps are deleted afterwards.

```python
import bootstrap  # noqa: F401 -- puts the repository on the import path

import zmart_drivers.zeiss.light_microscope.zen_api_grpc as zen

# The gateway's control token: copy it from the ZEN API Gateway's tray icon, or leave it
# out when this computer runs the gateway (the driver reads its token file then).
scope = zen.connect(password=None, experiment="ZMART_Snap")

from zmart_drivers.zeiss.light_microscope.zen_api_grpc.driver_setup.orientation import (
    measure_orientation as orientation,
)
```

## Measure

```python
measured = orientation.measure_orientation(scope.handle)
print(orientation.describe(measured))
```

## Adopt

Check the description above. When it is right, save it; every later connection loads it.

```python
adopted = orientation.adopt_orientation(measured)
scope.close()
adopted
```
