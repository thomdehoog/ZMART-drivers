# Set the limits of this FV4000

How far the stage and the focus may travel, and what each setting may be set to. The driver refuses every move outside these limits before anything is sent, so measure them with care. Ranges include both endpoints; an empty list means reviewed and unrestricted.

FLUOVIEW counts positions in nanometres from a zero of its own. Where that zero lies on the stage is not known until you measure it here, so the starting values below are placeholders, not this microscope's travel.

## Configure

Review the starting limits. The Measure step replaces X, Y and Z with what you record; the piezo range and the settings you edit by hand, if you want to bound them.

```python
import bootstrap

from zmart_drivers.evident.fv4000.rdk.driver_setup.limits.config import (
    adopt_limits,
    validate_limits,
)

LIMITS = {
    "x_um": {"range": [-50000, 50000]},
    "y_um": {"range": [-50000, 50000]},
    "z_um": {"range": [-5000, 5000]},
    "z_piezo_um": {"range": [-100, 100]},
    "objective_slot": [],
    "set_channel_enable": [],
    "set_channel_hv": [],
    "set_channel_gain": [],
    "set_channel_offset": [],
    "set_channel_digital_gain": [],
    "set_channel_laser": [],
    "set_laser_enable": [],
    "set_laser_intensity": [],
    "set_z_stack_enable": [],
    "set_z_stack_slices": [],
    "set_z_stack_step": [],
    "set_z_stack_start": [],
    "set_z_stack_end": [],
    "set_z_stack_range": [],
    "set_z_type": [],
    "set_time_lapse_enable": [],
    "set_time_lapse_repeat": [],
    "set_time_lapse_interval": [],
    "set_scanner_type": [],
    "set_scanner_mode": [],
    "set_load_parameter": [],
}
```

## Measure

Connect to FLUOVIEW. Then drive the stage to each safe corner of its travel, with the joystick or in FLUOVIEW, and run that corner's cell; it reads where the stage stands. Do the same for the lowest and highest safe focus position. To record a point again, move there and run its cell again.

```python
import zmart_drivers.evident.fv4000.rdk as fv4000
from zmart_drivers.evident.fv4000.rdk.driver_setup.limits.measure_limits import (
    TravelRecorder,
)

scope = fv4000.connect(load_calibration=False, load_origin=False)
travel = TravelRecorder(scope.client)
```

**Corner 1:** drive the stage to the first safe corner, then run the cell below.

```python
travel.record_corner(1)
```

**Corner 2:** drive the stage to the second safe corner, then run the cell below.

```python
travel.record_corner(2)
```

**Corner 3:** drive the stage to the third safe corner, then run the cell below.

```python
travel.record_corner(3)
```

**Corner 4:** drive the stage to the fourth safe corner, then run the cell below.

```python
travel.record_corner(4)
```

**Lowest focus:** move the focus to the lowest position that is safe for your samples, then run the cell below.

```python
travel.record_z("low")
```

**Highest focus:** move the focus to the highest safe position (watch the objective and the sample), then run the cell below.

```python
travel.record_z("high")
```

The X, Y and Z limits spanned by what you recorded.

```python
measured = travel.limits()
LIMITS.update(measured["limits"])
for axis in ("x_um", "y_um", "z_um"):
    print(axis, LIMITS[axis]["range"])
```

## Validate

Check the complete limits document before publishing it.

```python
validated_limits = validate_limits(LIMITS)
```

## Save and adopt

This saves the completed notebook, checks that it holds the measurements, then publishes the limits together with the notebook, which records the points you measured.

```python
# Save and Adopt
published = bootstrap.NOTEBOOK.save_and_adopt(
    lambda notebook: adopt_limits(validated_limits, notebook_paths=[notebook])
)
published
```
