# Set the orientation

How the picture is turned relative to the stage. In FLUOVIEW, set up a quick single-plane scan (one channel is enough) on a field with recognisable structure, and focus. This takes a picture here, one after a small move along X, and one after a small move along Y, compares them, and brings the stage back. Do the limits first: the moves go through them.

```python
import bootstrap

import zmart_drivers.evident.fv4000.rdk as fv4000
from zmart_drivers.evident.fv4000.rdk.driver_setup.orientation import (
    measure_orientation,
)

scope = fv4000.connect(load_calibration=False)
```

## Measure

Run the cell. It moves the stage about a quarter of the picture along X and along Y and back; pass `step_um=` to choose the distance yourself.

```python
result = measure_orientation.measure(scope.handle)
print(measure_orientation.describe(result))
```

## Save and adopt

This saves the completed notebook and publishes the measured orientation with it. A measurement that was not clear is refused, with the reason; move to a field with more structure and measure again.

```python
# Save and Adopt
adopted = bootstrap.NOTEBOOK.save_and_adopt(
    lambda notebook: measure_orientation.adopt_orientation(result, notebook)
)
adopted
```
