# Set the data location

The folder every run of this microscope is saved under, for example `D:\ZMART-data`: a full path to an existing folder on a **local disk** that this computer can write to. Copy the data to a network share after the run, not during it, because NIS waits while a picture is written.

It is set once, like the limits and the orientation, and the driver loads it at every connection. A session can still choose another folder (`output_root` when connecting), and so can a single capture (`acquire`'s `output_folder`).

Run the cells from the top. The first one shows where pictures go now.

```python
import bootstrap  # noqa: F401 -- puts the repository on the import path

from zmart_drivers.nikon.nis_elements_ar7.driver_setup import store
from zmart_drivers.nikon.nis_elements_ar7.driver_setup.data_location import (
    data_location,
    set_data_location,
)

store.use_microscope("nikon-microscope")  # this microscope's name, as in its connection
print("now:", data_location())
```

## Choose the folder

Write the folder below and run the cell. A folder that cannot hold the data is refused, with the reason.

```python
set_data_location(r"D:\ZMART-data")
print("set:", data_location())
```
