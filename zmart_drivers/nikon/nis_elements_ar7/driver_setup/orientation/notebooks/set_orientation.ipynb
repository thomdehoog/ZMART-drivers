# Set the orientation of the picture to the stage

A camera is often mounted a quarter turn, or a mirror image, away from the stage's own x and y. Then moving the stage "right" moves the sample in some other direction in the picture. The driver turns every saved picture so that its left to right is the stage's +x and its top to bottom the stage's +y, so that a workflow can treat picture and stage directions as one. It only rearranges pixels: it never blurs or resamples the picture.

This notebook asks you to state the turn by looking: you take a picture, move the stage by +x and take another, then by +y and take a third, and see which way the sample moved.

You need NIS-Elements running, with `start_bridge.mac` started in it.

```python
from zmart_drivers.nikon import nis_elements_ar7
from zmart_drivers.nikon.nis_elements_ar7.driver_setup.orientation import (
    Orientation,
    save_orientation,
)

scope = nis_elements_ar7.connect(
    microscope="nikon-microscope"
)  # this microscope's name
scope.position()
```

## Look

Focus on a sample with a clear feature. Take a picture in NIS, then run the next cell, which moves the stage 20 µm in +x; take a second picture. Then run the cell after it, which moves 20 µm in +y; take a third. Compare where the feature went.

```python
here = scope.position()
scope.move_to(here["x"]["position"] + 20, here["y"]["position"], here["z"]["position"])
```

```python
scope.move_to(
    here["x"]["position"] + 20, here["y"]["position"] + 20, here["z"]["position"]
)
```

## State the turn

When +x moved the feature to the **left** and +y moved it **up** in the picture, the picture is already lined up with the stage: `rotate_deg=0, mirrored=False`. (The feature moves the opposite way to the stage.) Otherwise find the quarter turn, clockwise, and whether a mirror is needed, that brings the picture to that case. Then run the cell.

```python
save_orientation(Orientation(rotate_deg=0, mirrored=False))
```

```python
scope.move_to(here["x"]["position"], here["y"]["position"], here["z"]["position"])
scope.close()
```
