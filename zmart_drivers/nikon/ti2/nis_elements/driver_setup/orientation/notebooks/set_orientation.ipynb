# Set the camera orientation

Which way the camera is turned relative to the stage, so that saved pictures follow the stage's x and y. The driver steps the stage a little along x and then along y, compares the pictures, and finds the turn. Put a sample with some features in view and bring it into focus first; a blank field gives no answer.

```python
import bootstrap

from zmart_drivers.nikon.ti2 import nis_elements
from zmart_drivers.nikon.ti2.nis_elements.driver_setup.orientation import (
    adopt_orientation,
)
from zmart_drivers.nikon.ti2.nis_elements.driver_setup.orientation.measure_orientation import (
    measure,
)

scope = nis_elements.connect(load_calibration=False)
print(scope.position())
```

## Measure

The stage steps 50 micrometres along x and along y, and comes back. Make sure that much travel is free on both sides.

```python
measured = measure(scope.handle, step_um=50.0)
print("turn:", measured["orientation"])
print("pixel size on the sample:", round(measured["pixel_size_um"], 4), "um")
print("how clear the shifts were:", measured["peak_ratios"])
```

## Save and adopt

This saves the completed notebook, then publishes the orientation together with it. Every later connect applies the turn to the pictures it saves.

```python
published = bootstrap.NOTEBOOK.save_and_adopt(
    lambda notebook: adopt_orientation(
        measured["orientation"],
        pixel_size_um=measured["pixel_size_um"],
        archive_paths=[notebook],
    )
)
published
```
