# Calibrate a pair of objectives

How far the sample shifts, in x, y and focus, when the objective changes. With it, the driver keeps the same spot in view when it changes objectives. You centre the same small feature under each objective; the driver records where the stage stands and takes the difference.

```python
import bootstrap

from zmart_drivers.nikon.ti2 import nis_elements
from zmart_drivers.nikon.ti2.nis_elements.driver_setup.optical_calibration.calibrate_objective_pair import (
    ObjectivePairRecorder,
)

scope = nis_elements.connect(load_calibration=False)
print("objectives:", scope.objectives())
pair = ObjectivePairRecorder(scope.handle)
```

## The first objective

With the first objective in place, centre a small feature in the picture and bring it into focus (joystick, or NIS-Elements). Then record.

```python
pair.record()
```

## The second objective

Change to the second objective: write its nosepiece slot below. The driver turns the nosepiece and moves nothing else. Then centre and focus the same feature again, and record.

```python
pair.change_objective(2)
```

```python
pair.record()
```

The offsets, relative to the reference objective.

```python
pair.translations()
```

## Save and adopt

This saves the completed notebook, then publishes the offsets together with it, over the pairs measured before. Every later connect loads them.

```python
published = bootstrap.NOTEBOOK.save_and_adopt(
    lambda notebook: pair.adopt(notebook_paths=[notebook])
)
published
```
