# Set microscope limits

Set the safe movement and hardware ranges for this microscope. Ranges include both endpoints; an empty list means reviewed and unrestricted.

## Configure

Review the starting limits. Z-wide starts at zero because it cannot move to a negative position; the Measure step replaces only X and Y with the rectangle spanned by the four recorded corners.

```python
import _bootstrap

from zmart_drivers.leica.stellaris5_y42h93.navigator_expert.configuration.limits.config import (
    adopt_limits,
    validate_limits,
)

LIMITS = {
    "x_um": {"range": [1000, 130000]},
    "y_um": {"range": [1000, 100000]},
    "z_galvo_um": {"range": [-250, 250]},
    "z_wide_um": {"range": [0, 8000]},
    "objective_slot": [],
    "set_zoom": [],
    "set_scan_speed": [],
    "set_scan_resonant": [],
    "set_scan_mode": [],
    "set_sequential_mode": [],
    "set_scan_field_rotation": [],
    "set_image_format": [],
    "set_z_stack_definition": [],
    "set_z_stack_step_size": [],
    "set_z_stack_size": [],
    "set_frame_accumulation": [],
    "set_frame_average": [],
    "set_line_accumulation": [],
    "set_line_average": [],
    "set_pinhole_airy": [],
    "set_detector_gain": [],
    "set_laser_intensity": [],
    "set_laser_shutter": [],
    "set_filter_wheel_slot": [],
    "set_filter_wheel_spectrum": [],
}
```

## Measure

Record the four corners of the stage's safe travel. Drive the stage to each corner with the joystick or in LAS X, then run that corner's cell; it reads where the stage stands. To record a corner again, move the stage and run its cell again. Never make positions in the Navigator Expert.

```python
import zmart_drivers.leica.stellaris5_y42h93.navigator_expert as drv
from zmart_drivers.leica.stellaris5_y42h93.navigator_expert.procedures.measure_limits import (
    CornerRecorder,
)

client = drv.connect_microscope(load_calibration=False)
corners = CornerRecorder(client)
```

**Corner 1:** drive the stage to the first safe corner, then run the cell below.

```python
corners.record(1)
```

**Corner 2:** drive the stage to the second safe corner, then run the cell below.

```python
corners.record(2)
```

**Corner 3:** drive the stage to the third safe corner, then run the cell below.

```python
corners.record(3)
```

**Corner 4:** drive the stage to the fourth safe corner, then run the cell below.

```python
corners.record(4)
```

The X and Y limits spanned by the four corners.

```python
captured_xy = corners.limits()
LIMITS.update(captured_xy["limits"])
print("X:", LIMITS["x_um"]["range"])
print("Y:", LIMITS["y_um"]["range"])
```

## Validate

Check the complete limits document before saving or adopting it.

```python
validated_limits = validate_limits(LIMITS)
```

## Save and Adopt

This saves the completed notebook, verifies its measurement and validation outputs, then publishes the limits with the notebook, which records the four corners.

```python
# Save and Adopt
published = _bootstrap.NOTEBOOK.save_and_adopt(
    lambda notebook: adopt_limits(
        validated_limits,
        notebook_paths=[notebook],
    )
)
published
```
